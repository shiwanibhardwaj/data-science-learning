```python
import pandas as pd

data=pd.read_csv("D:\\shivani\\data science\\pandas\\E-commerce_Sales\\customers.csv")
data

data.head(10)
data.tail(10)
data.info()
data.describe()
data.isnull().sum()
data.duplicated().sum()

new_data=data.copy()
new_data["Age"]=new_data["Age"].fillna(new_data["Age"].mean())
print(new_data.isnull().sum())

new_data["City"]=new_data["City"].fillna("Unknown")
print(new_data.isnull().sum())
new_data

new_data["SignupDate"]=pd.to_datetime(new_data["SignupDate"])
new_data.info()

new_data["City"].value_counts()
new_data["CustomerSegment"].value_counts()
```

```python
new_data=new_data.assign(
    SignupYear=new_data["SignupDate"].dt.year,
    SignupMonth=new_data["SignupDate"].dt.month,
    SignupDay=new_data["SignupDate"].dt.day,
    SignupDayOfWeek=new_data["SignupDate"].dt.day_name()
)
new_data
```

```python
avg_age=new_data.groupby("CustomerSegment")["Age"].agg(avg_age="mean")
avg_age

city_count=new_data.groupby('City')['CustomerID'].count()
city_count

avg_city_age=new_data.groupby('City')['Age'].mean()
avg_city_age
```

```python
data2=pd.read_csv("D:\\shivani\\data science\\pandas\\E-commerce_Sales\\orders.csv")
order_data=data2.copy()
order_data.info()
order_data.isnull().sum()
order_data["PaymentMethod"]=order_data["PaymentMethod"].fillna("Not Specified")
order_data["PaymentMethod"]
order_data["Discount"] = order_data["Discount"].fillna(
    order_data["Discount"].median()
)


order_data["Quantity"].describe()
(order_data["Quantity"] < 0).sum()
order_data[order_data["Quantity"] < 0][["OrderID", "CustomerID", "Quantity", "Status"]]
order_data[order_data["Quantity"] < 0]["Quantity"].value_counts()
order_data.loc[order_data["Quantity"]<0,"Quantity"]=pd.NA
order_data["Quantity"].isnull().sum()
order_data["Quantity"]=order_data["Quantity"].fillna(order_data["Quantity"].median())
order_data["Quantity"].isnull().sum()


order_data["OrderDate"]=pd.to_datetime(order_data["OrderDate"],errors="coerce")
order_data["OrderDate"]
order_data["OrderDate"].isnull().sum()
order_data[order_data["OrderDate"].isnull()][["OrderID", "CustomerID", "Quantity", "Status"]]

#Check whether the same CustomerID has other orders with known dates
missing_dates=order_data[order_data["OrderDate"].isna()]
order_data[
    order_data["CustomerID"].isin(missing_dates["CustomerID"])][["OrderID", "CustomerID","OrderDate","Quantity", "Status"]].sort_values(["CustomerID", "OrderDate"])

print(missing_dates["CustomerID"].value_counts())
missing_dates[["OrderID", "CustomerID", "ProductID", "Quantity", "Status"]]

#OrderDate: 35 values were missing.Since the actual order dates could not be reliably determined from the available data, 
#the missing values were retained as NaT rather than assigning arbitrary dates.
```

```python
order_ids=set(order_data["OrderID"])
payments_order_ids=set(payments_data["OrderID"])
missing_payment_orders=order_ids-payments_order_ids

len(missing_payment_orders)

missing_order_dates=order_data[order_data["OrderDate"].isna()]
missing_payment_dates=payments_data[payments_data["PaymentDate"].isna()]

missing_order_dates["OrderID"].isin(missing_payment_dates["OrderID"]).sum()
#The 35 missing order dates and 35 missing payment dates correspond to the same orders.
#Since no reliable source is available to reconstruct those dates, they were retained as missing.
```

```python
# merging Order + Customer
order_customer=order_data.merge(
    new_data,
    on="CustomerID",
    how="left"
)
order_customer.shape
order_customer["CustomerID"].isna().sum()
order_customer
```

```python
# merging Order + Customer + products
order_customer_product=order_customer.merge(
    products_data,
    on="ProductID",
    how="left"
)
order_customer_product
```

```python
#merging Order + Customer + products + payments
full_data = order_customer_product.merge(
    payments_data,
    on="OrderID",
    how="left"
)
full_data
```

```python
#1. Sales & Order Analysis

#How many total orders are there?
full_data["OrderID"].nunique()
#Which month has the most orders?
full_data.groupby(full_data["OrderDate"].dt.month)["OrderID"].nunique().idxmax()
#Most common order status
full_data["Status"].value_counts()
```

```python
#Customer Analysis

#Which customer segment places the most orders?
full_data.groupby(full_data["CustomerSegment"])["OrderID"].nunique().idxmax()
```
