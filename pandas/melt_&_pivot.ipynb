```python
import pandas as pd
#melt
data=pd.read_csv("student_data.csv")
data=data.melt(
    id_vars=["Student_ID","Name","Age","Gender","Attendance","City"],
    var_name="Subjects",
    value_name="Scores"
)
data


#pivot
data=data.pivot(
    index=[
    "Student_ID", "Name", "Age",
    "Gender", "Attendance"],
    columns="Subjects",
    values="Scores"
)
print(data)
```
