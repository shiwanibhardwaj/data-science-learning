```python
import pandas as pd

df1 = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Aman"],
    "Marks": [85, 78, 92]
})

df2 = pd.DataFrame({
    "Student_ID": [104, 105, 106],
    "Name": ["Neha", "Rahul", "Sneha"],
    "Marks": [88, 76, 95]
})
result=pd.concat([df1,df2])
print(result)

result=pd.concat([df1,df2],ignore_index=True,axis=1) # ignore_index=True - for ignoring the repeated index.
result
```
