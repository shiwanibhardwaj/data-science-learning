```python
import pandas as pd
data=pd.read_json("students_data.json")
```

```python
data.isnull()
data.isnull().sum()
data.isna()

data.dropna() #row drop
data.dropna(axis=1) #col drop

data.fillna("shivani")
data["Age"].fillna(data["Age"].mean())


#these methods doesn't make changes in original data ,thus, we create the copy of original data and work on it.
cleaned_data=data.copy()
age_mean=cleaned_data["Age"].mean()
cleaned_data["Age"]=cleaned_data["Age"].fillna(age_mean)
cleaned_data

#ffill and bfill (forward fill and backward fill)
cleaned_data.ffill()
cleaned_data.bfill()
cleaned_data.dtypes
```

```python
#handle duplicate values
data.duplicated()
data["Name"].duplicated()
data.drop_duplicates(inplace=True)
data
```

```python
# datetime
date_str="2026-12-31"
print(date_str)
s=pd.to_datetime(date_str)
data["Join_Date"]=data["Join_Date"].fillna(s)
```

```python
#handle strings
data["Name"].str.capitalize()
data["Name"].str.upper()
data["Name"].str.lower()
data["Name"].str.strip()
data["Name"].str.split(" ")
data["Email"].str.split("@")
data["Name"].str.contains("Simran Kaur",case=False)
```
