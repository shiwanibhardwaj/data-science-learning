```python
import pandas as pd
#grouping and aggregation
data=pd.read_csv("titanic/train.csv")
data2=data.copy()
data2.groupby("Sex")["Age"].mean()
data2.groupby("Sex")["Age"].min()
data2.groupby("Sex")["Age"].max()
#agg()
print(data2.groupby("Sex")["Age"].agg(["mean","max","min","median"])) #agg() or aggregate()

#passing new lables
data2.groupby("Sex")["Age"].agg(avg_age="mean",max_age="max",min_age="min",mid_age="median")


data2.groupby("Sex").agg({
    "Age":"mean",
    "Pclass":"mean"
    
})
#passing new lables
data2.groupby("Sex").agg(
    avg_age=("Age","mean"),
    avg_pclass=("Pclass","mean")
    
)
```
