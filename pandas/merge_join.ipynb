```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Aman"],
    "City": ["Delhi", "Mumbai", "Pune"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 104],
    "Python": [85, 78, 92],
    "SQL": [90, 88, 95]
})
result=pd.merge(students,marks,on="Student_ID")
result  #by default inner merge.


#inner merge=Only matching IDs
result=pd.merge(students,marks,on="Student_ID",how="inner")
result

#left-LEFT = keep everything from left DataFrame.
result=pd.merge(students,marks,on="Student_ID",how="left")
result

#left-LEFT = keep everything from right DataFrame.
result=pd.merge(students,marks,on="Student_ID",how="right")
result

#outer = Keeps everything from both DataFrames.
result=pd.merge(students,marks,on="Student_ID",how="outer")
result


# if column names are different
students = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Aman"],
    "City": ["Delhi", "Mumbai", "Pune"]
})

marks = pd.DataFrame({
    "Marks_ID": [101, 102, 104],
    "Python": [85, 78, 92],
    "SQL": [90, 88, 95]
})
result=pd.merge(students,marks,left_on="Student_ID",right_on="Marks_ID",how="outer")
result
```
