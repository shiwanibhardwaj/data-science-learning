```python
import pandas as pd
```

```python
#apply() is useful when you want to perform a custom transformation.
data={
    "Names":["shivani","abhishek","saurabh"],
    "Age":[20,26,28],
    "salary":[20000,50000,70000]
}
df=pd.DataFrame(data)

def age_group(Age):
    if Age<25:
        return "young"
    else:
        return "adult"

df["Age_group"]=df["Age"].apply(age_group)
print(df["Age_group"])


#apply method with lambda...
df["increased_saraly"]=df["salary"].apply(lambda x:x + 1000 if x<60000 else x)
print(df["increased_saraly"])
print(df)
```

```python
#map() is especially useful for categorical data.
data={
    "Names":["shivani","abhishek","saurabh"],
    "Age":[20,26,28],
    "salary":[20000,50000,70000],
    "gender":["female","male","male"]
}
df=pd.DataFrame(data)

df["gender_code"]=df["gender"].map({
    "female":1,
    "male":0
})
print(df["gender_code"])
print(df)
```

```python
#assign- (is useful for creating new features while keeping your DataFrame transformation clean.)

data={
    "Names":["shivani","abhishek","saurabh"],
    "Age":[20,26,28],
    "salary":[20000,50000,70000],
    "gender":["female","male","male"]
}
df=pd.DataFrame(data)
df=df.assign(
    gender_code=df["gender"].map({"female":0,"male":1})
)
print(df)
```

```python
#replace
data={
    "Names":["shivani","abhishek","saurabh"],
    "Age":[20,26,28],
    "salary":[20000,50000,70000],
    "gender":["female","male","male"]
}
df=pd.DataFrame(data)

df["Names"]=df["Names"].replace("shivani","shivani kumari")
print(df)
```

```python
#rename
df.rename(columns={"salary":"Income"})
df.rename(columns={"gender":"Gender"})

df.rename(index={1:"first"})
```
